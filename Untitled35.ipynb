```python
# ============================================================
# COMPLETE PYTHON ANALYSIS
# ============================================================

import pandas as pd
import numpy as np

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix
)

pd.set_option("display.max_columns", None)
pd.set_option("display.max_rows", 100)


# ============================================================
# 1. LOAD CLEANED DATA
# ============================================================

df = pd.read_csv("Cleaned_Dataset.csv")

print("=" * 60)
print("DATASET LOADED")
print("=" * 60)

print("Rows:", df.shape[0])
print("Columns:", df.shape[1])

display(df.head())


# ============================================================
# 2. BASIC DATA INFORMATION
# ============================================================

print("\n" + "=" * 60)
print("DATA TYPES")
print("=" * 60)

print(df.dtypes)


print("\n" + "=" * 60)
print("MISSING VALUES")
print("=" * 60)

print(df.isnull().sum())


print("\n" + "=" * 60)
print("DUPLICATE RECORDS")
print("=" * 60)

print("Duplicates:", df.duplicated().sum())


# ============================================================
# 3. DATE CONVERSION
# ============================================================

df["Event_Date"] = pd.to_datetime(
    df["Event_Date"],
    errors="coerce"
)

print("\nDate Range:")
print("Start Date:", df["Event_Date"].min())
print("End Date:", df["Event_Date"].max())


# ============================================================
# 4. DESCRIPTIVE STATISTICS
# ============================================================

print("\n" + "=" * 60)
print("DESCRIPTIVE STATISTICS")
print("=" * 60)

numeric_columns = [
    "Prediction_Score",
    "Latency_ms",
    "Inference_Cost_USD",
    "Drift_Score",
    "Fairness_Gap",
    "Business_Value_USD"
]

display(df[numeric_columns].describe())


# ============================================================
# 5. MAIN BUSINESS & MODEL KPIs
# ============================================================

total_records = len(df)

total_business_value = df["Business_Value_USD"].sum()

average_business_value = df["Business_Value_USD"].mean()

average_latency = df["Latency_ms"].mean()

average_prediction_score = df["Prediction_Score"].mean()

average_drift = df["Drift_Score"].mean()

average_fairness_gap = df["Fairness_Gap"].mean()

total_inference_cost = df["Inference_Cost_USD"].sum()

average_inference_cost = df["Inference_Cost_USD"].mean()


print("\n" + "=" * 60)
print("KEY PERFORMANCE INDICATORS")
print("=" * 60)

print("Total Records:", total_records)
print("Total Business Value: $", round(total_business_value, 2))
print("Average Business Value: $", round(average_business_value, 2))
print("Average Latency (ms):", round(average_latency, 2))
print("Average Prediction Score:", round(average_prediction_score, 3))
print("Average Drift Score:", round(average_drift, 3))
print("Average Fairness Gap:", round(average_fairness_gap, 3))
print("Total Inference Cost: $", round(total_inference_cost, 2))
print("Average Inference Cost: $", round(average_inference_cost, 4))


# ============================================================
# 6. MODEL ACCURACY
# ============================================================

actual = df["Actual_Class"]
predicted = df["Predicted_Class"]

accuracy = accuracy_score(actual, predicted)

precision = precision_score(
    actual,
    predicted,
    pos_label="Positive",
    zero_division=0
)

recall = recall_score(
    actual,
    predicted,
    pos_label="Positive",
    zero_division=0
)

f1 = f1_score(
    actual,
    predicted,
    pos_label="Positive",
    zero_division=0
)


print("\n" + "=" * 60)
print("MODEL PERFORMANCE")
print("=" * 60)

print("Accuracy :", round(accuracy, 4))
print("Accuracy %:", round(accuracy * 100, 2))
print("Precision:", round(precision, 4))
print("Recall   :", round(recall, 4))
print("F1 Score :", round(f1, 4))


# ============================================================
# 7. CONFUSION MATRIX
# ============================================================

cm = confusion_matrix(
    actual,
    predicted,
    labels=["Positive", "Negative"]
)

tn_positive = cm[1, 1]
fp_positive = cm[1, 0]
fn_positive = cm[0, 1]
tp_positive = cm[0, 0]

print("\n" + "=" * 60)
print("CONFUSION MATRIX")
print("=" * 60)

print("Matrix:")
print(cm)

print("\nTrue Positive :", tp_positive)
print("True Negative :", tn_positive)
print("False Positive:", fp_positive)
print("False Negative:", fn_positive)


# ============================================================
# 8. MODEL-WISE PERFORMANCE
# ============================================================

model_performance = df.groupby("Model_Name").agg(
    Records=("Event_ID", "count"),
    Avg_Prediction_Score=("Prediction_Score", "mean"),
    Avg_Latency_ms=("Latency_ms", "mean"),
    Avg_Drift_Score=("Drift_Score", "mean"),
    Avg_Fairness_Gap=("Fairness_Gap", "mean"),
    Total_Cost_USD=("Inference_Cost_USD", "sum"),
    Total_Business_Value_USD=("Business_Value_USD", "sum")
).reset_index()


print("\n" + "=" * 60)
print("MODEL-WISE ANALYSIS")
print("=" * 60)

display(model_performance)


# ============================================================
# 9. REGION-WISE ANALYSIS
# ============================================================

region_analysis = df.groupby("Region").agg(
    Records=("Event_ID", "count"),
    Avg_Latency_ms=("Latency_ms", "mean"),
    Avg_Drift_Score=("Drift_Score", "mean"),
    Avg_Fairness_Gap=("Fairness_Gap", "mean"),
    Total_Cost_USD=("Inference_Cost_USD", "sum"),
    Total_Business_Value_USD=("Business_Value_USD", "sum")
).reset_index()


print("\n" + "=" * 60)
print("REGION-WISE ANALYSIS")
print("=" * 60)

display(region_analysis)


# ============================================================
# 10. OUTCOME ANALYSIS
# ============================================================

outcome_analysis = df.groupby("Outcome_Status").agg(
    Records=("Event_ID", "count"),
    Total_Business_Value=("Business_Value_USD", "sum"),
    Average_Business_Value=("Business_Value_USD", "mean"),
    Average_Latency=("Latency_ms", "mean"),
    Average_Drift=("Drift_Score", "mean")
).reset_index()


print("\n" + "=" * 60)
print("OUTCOME ANALYSIS")
print("=" * 60)

display(outcome_analysis)


# ============================================================
# 11. SUCCESS / FAILURE RATE
# ============================================================

success_count = (
    df["Outcome_Status"] == "Success"
).sum()

failure_count = (
    df["Outcome_Status"] == "Failure"
).sum()

success_rate = success_count / total_records
failure_rate = failure_count / total_records


print("\n" + "=" * 60)
print("OUTCOME RATES")
print("=" * 60)

print("Success Count:", success_count)
print("Failure Count:", failure_count)

print("Success Rate:", round(success_rate * 100, 2), "%")
print("Failure Rate:", round(failure_rate * 100, 2), "%")


# ============================================================
# 12. DRIFT ANALYSIS
# ============================================================

drift_threshold = 0.20

df["Drift_Status"] = np.where(
    df["Drift_Score"] >= drift_threshold,
    "Alert",
    "Normal"
)

drift_alerts = (
    df["Drift_Status"] == "Alert"
).sum()

drift_alert_rate = drift_alerts / total_records


print("\n" + "=" * 60)
print("DRIFT ANALYSIS")
print("=" * 60)

print("Average Drift Score:",
      round(df["Drift_Score"].mean(), 3))

print("Maximum Drift Score:",
      round(df["Drift_Score"].max(), 3))

print("Drift Alerts:", drift_alerts)

print("Drift Alert Rate:",
      round(drift_alert_rate * 100, 2), "%")


# ============================================================
# 13. FAIRNESS ANALYSIS
# ============================================================

fairness_threshold = 0.10

df["Fairness_Status"] = np.where(
    df["Fairness_Gap"] >= fairness_threshold,
    "Alert",
    "Normal"
)

fairness_alerts = (
    df["Fairness_Status"] == "Alert"
).sum()

fairness_alert_rate = fairness_alerts / total_records


print("\n" + "=" * 60)
print("FAIRNESS ANALYSIS")
print("=" * 60)

print("Average Fairness Gap:",
      round(df["Fairness_Gap"].mean(), 3))

print("Maximum Fairness Gap:",
      round(df["Fairness_Gap"].max(), 3))

print("Fairness Alerts:", fairness_alerts)

print("Fairness Alert Rate:",
      round(fairness_alert_rate * 100, 2), "%")


# ============================================================
# 14. LATENCY ANALYSIS
# ============================================================

latency_threshold = 300

df["Latency_Status"] = np.where(
    df["Latency_ms"] >= latency_threshold,
    "Alert",
    "Normal"
)

latency_alerts = (
    df["Latency_Status"] == "Alert"
).sum()

latency_alert_rate = latency_alerts / total_records


print("\n" + "=" * 60)
print("LATENCY ANALYSIS")
print("=" * 60)

print("Average Latency:",
      round(df["Latency_ms"].mean(), 2), "ms")

print("Maximum Latency:",
      round(df["Latency_ms"].max(), 2), "ms")

print("Latency Alerts:", latency_alerts)

print("Latency Alert Rate:",
      round(latency_alert_rate * 100, 2), "%")


# ============================================================
# 15. COST ANALYSIS
# ============================================================

print("\n" + "=" * 60)
print("COST ANALYSIS")
print("=" * 60)

print("Total Inference Cost: $",
      round(df["Inference_Cost_USD"].sum(), 2))

print("Average Cost per Prediction: $",
      round(df["Inference_Cost_USD"].mean(), 4))

print("Maximum Cost per Prediction: $",
      round(df["Inference_Cost_USD"].max(), 4))


# Cost per Business Value
cost_to_value_ratio = (
    df["Inference_Cost_USD"].sum() /
    df["Business_Value_USD"].sum()
)

print("Cost / Business Value Ratio:",
      round(cost_to_value_ratio, 6))


# ============================================================
# 16. BUSINESS VALUE ANALYSIS
# ============================================================

print("\n" + "=" * 60)
print("BUSINESS VALUE ANALYSIS")
print("=" * 60)

print("Total Business Value: $",
      round(df["Business_Value_USD"].sum(), 2))

print("Average Business Value: $",
      round(df["Business_Value_USD"].mean(), 2))

print("Highest Business Value: $",
      round(df["Business_Value_USD"].max(), 2))

print("Lowest Business Value: $",
      round(df["Business_Value_USD"].min(), 2))


# ============================================================
# 17. MODEL HEALTH
# ============================================================

df["Model_Health"] = np.where(
    (df["Drift_Score"] < 0.20) &
    (df["Fairness_Gap"] < 0.10) &
    (df["Latency_ms"] < 300),
    "Healthy",
    "Needs Attention"
)

health_summary = df["Model_Health"].value_counts()

print("\n" + "=" * 60)
print("MODEL HEALTH")
print("=" * 60)

print(health_summary)


# ============================================================
# 18. ALERT SUMMARY
# ============================================================

alert_summary = pd.DataFrame({
    "Metric": [
        "Drift",
        "Fairness",
        "Latency"
    ],
    "Alert_Count": [
        drift_alerts,
        fairness_alerts,
        latency_alerts
    ],
    "Alert_Rate_%": [
        drift_alert_rate * 100,
        fairness_alert_rate * 100,
        latency_alert_rate * 100
    ]
})

alert_summary["Alert_Rate_%"] = alert_summary[
    "Alert_Rate_%"
].round(2)


print("\n" + "=" * 60)
print("ALERT SUMMARY")
print("=" * 60)

display(alert_summary)


# ============================================================
# 19. MODEL HEALTH BY MODEL
# ============================================================

model_health = pd.crosstab(
    df["Model_Name"],
    df["Model_Health"]
)

print("\n" + "=" * 60)
print("MODEL HEALTH BY MODEL")
print("=" * 60)

display(model_health)


# ============================================================
# 20. CORRELATION ANALYSIS
# ============================================================

correlation_columns = [
    "Prediction_Score",
    "Latency_ms",
    "Inference_Cost_USD",
    "Drift_Score",
    "Fairness_Gap",
    "Business_Value_USD"
]

correlation_matrix = df[
    correlation_columns
].corr()

print("\n" + "=" * 60)
print("CORRELATION ANALYSIS")
print("=" * 60)

display(correlation_matrix.round(3))


# ============================================================
# 21. HIGHEST / LOWEST PERFORMING MODELS
# ============================================================

best_model_business = model_performance.loc[
    model_performance["Total_Business_Value_USD"].idxmax()
]

lowest_model_business = model_performance.loc[
    model_performance["Total_Business_Value_USD"].idxmin()
]

highest_drift_model = model_performance.loc[
    model_performance["Avg_Drift_Score"].idxmax()
]

highest_latency_model = model_performance.loc[
    model_performance["Avg_Latency_ms"].idxmax()
]


print("\n" + "=" * 60)
print("KEY MODEL INSIGHTS")
print("=" * 60)

print(
    "Highest Business Value Model:",
    best_model_business["Model_Name"]
)

print(
    "Lowest Business Value Model:",
    lowest_model_business["Model_Name"]
)

print(
    "Highest Average Drift Model:",
    highest_drift_model["Model_Name"]
)

print(
    "Highest Average Latency Model:",
    highest_latency_model["Model_Name"]
)


# ============================================================
# 22. AUTOMATIC INSIGHT SUMMARY
# ============================================================

print("\n" + "=" * 60)
print("AUTOMATIC INSIGHT SUMMARY")
print("=" * 60)

print(
    f"1. The dataset contains {total_records} cleaned records."
)

print(
    f"2. Overall model accuracy is {accuracy * 100:.2f}%."
)

print(
    f"3. Overall success rate is {success_rate * 100:.2f}%."
)

print(
    f"4. Average latency is {average_latency:.2f} ms."
)

print(
    f"5. Average drift score is {average_drift:.3f}."
)

print(
    f"6. Average fairness gap is {average_fairness_gap:.3f}."
)

print(
    f"7. Total inference cost is ${total_inference_cost:.2f}."
)

print(
    f"8. Total business value is ${total_business_value:.2f}."
)

print(
    f"9. Drift alert rate is {drift_alert_rate * 100:.2f}%."
)

print(
    f"10. Fairness alert rate is {fairness_alert_rate * 100:.2f}%."
)

print(
    f"11. Latency alert rate is {latency_alert_rate * 100:.2f}%."
)


# ============================================================
# 23. FINAL ANALYSIS TABLE
# ============================================================

final_kpis = pd.DataFrame({
    "Metric": [
        "Total Records",
        "Accuracy %",
        "Precision",
        "Recall",
        "F1 Score",
        "Success Rate %",
        "Failure Rate %",
        "Average Latency (ms)",
        "Average Drift Score",
        "Average Fairness Gap",
        "Total Inference Cost (USD)",
        "Average Inference Cost (USD)",
        "Total Business Value (USD)",
        "Average Business Value (USD)",
        "Drift Alert Rate %",
        "Fairness Alert Rate %",
        "Latency Alert Rate %"
    ],
    "Value": [
        total_records,
        round(accuracy * 100, 2),
        round(precision, 4),
        round(recall, 4),
        round(f1, 4),
        round(success_rate * 100, 2),
        round(failure_rate * 100, 2),
        round(average_latency, 2),
        round(average_drift, 3),
        round(average_fairness_gap, 3),
        round(total_inference_cost, 2),
        round(average_inference_cost, 4),
        round(total_business_value, 2),
        round(average_business_value, 2),
        round(drift_alert_rate * 100, 2),
        round(fairness_alert_rate * 100, 2),
        round(latency_alert_rate * 100, 2)
    ]
})


print("\n" + "=" * 60)
print("FINAL KPI TABLE")
print("=" * 60)

display(final_kpis)


# ============================================================
# 24. EXPORT PYTHON ANALYSIS RESULTS
# ============================================================

with pd.ExcelWriter(
    "Python_Analysis_Output.xlsx",
    engine="openpyxl"
) as writer:

    final_kpis.to_excel(
        writer,
        sheet_name="Final_KPIs",
        index=False
    )

    model_performance.to_excel(
        writer,
        sheet_name="Model_Performance",
        index=False
    )

    region_analysis.to_excel(
        writer,
        sheet_name="Region_Analysis",
        index=False
    )

    outcome_analysis.to_excel(
        writer,
        sheet_name="Outcome_Analysis",
        index=False
    )

    alert_summary.to_excel(
        writer,
        sheet_name="Alert_Summary",
        index=False
    )

    correlation_matrix.to_excel(
        writer,
        sheet_name="Correlation"
    )

    model_health.to_excel(
        writer,
        sheet_name="Model_Health"
    )


print("\n" + "=" * 60)
print("ANALYSIS COMPLETED")
print("=" * 60)

print("Output file created:")
print("Python_Analysis_Output.xlsx")
```
